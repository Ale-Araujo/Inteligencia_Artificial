## Gere uma matriz 4x4 com números inteiros de 1 a 16 e extraia a última linha

```python
import numpy as np #Usamos o numpy e batizamos para "np" para facilitar o uso de suas funções

matriz = np.arange(1, 17).reshape(4, 4)
#Usamos uma das funcionalidades do np: Arange, ela permite formar os numeros de 1 a 16 numeros(OBS: o N° 17 nao se inclui)
#E usamos o "reshape(4, 4)" para organizar a nossa matriz de 4x4, totalizando 16 numeros no total
print("Matriz Completa:")
print(matriz)

#A ultima linha, tecnicamente representado por "(-1) seleciona a ultima linha da matriz, assim como o nome sugere"
#Já o "(:) significa todas as colunas"

#Ou seja, a variavel "ultima_linha" seleciona a ultima linha e todas as colunas existentes
ultima_linha = matriz[-1, :]

#por fim, imprimimos a matriz
#(OBS: "\n" faz uma quebra de linha na hora de imprimir o codigo)
print("\nUltima Linha:")
print(ultima_linha)


#===============================================================================
#justificativa:

# Neste exercício, escolhi utilizar o NumPy porque ele facilita a criação e
# manipulação de matrizes. Usei o arange() para gerar os valores que seriam
# utilizados na matriz e o reshape() para organizá-los em uma estrutura 4x4.
# Para obter somente a última linha, utilizei a indexação [-1, :], pois ela
# permite selecionar diretamente a última linha e todas as suas colunas.
```

## Crie um array com 10 numeros aleatorios e normalize seus valores entre 0 e 1.

```python
import numpy as np

# Criamos um array com 10 números aleatórios entre 0 e 1
array = np.random.rand(10)


array_normalizado = (array - array.min()) / (array.max() - array.min())
# Usamos o "min" para encontrar o menor valor do array
# E usamos o "max" para encontrar o maior valor do array
# Depois, aplicamos uma fórmula para normalizar os valores entre 0 e 1

print("Array original:")
print(array)

print("\n Array normalizado:")
print(array_normalizado)

#===============================================================================
#justificativa:

# A normalização é aplicada para colocar todos os valores do array dentro de
# uma mesma escala, facilitando sua comparação. Para isso, utilizo os valores
# mínimo e máximo como limites e ajusto os demais valores proporcionalmente
# para que o menor represente 0 e o maior represente 1.
```

## Gere um array de 100 numeros inteiros e encontre o valor que mais se repete no array.

```python
import numpy as np

array = np.random.randint(1, 101, 100)
# Criamos um array com 100 números inteiros aleatórios entre 1 e 100

valores, contagem = np.unique(array, return_counts=True)
# Usamos o "unique" para encontrar os valores únicos do array
# O "return_counts=True" também retorna quantas vezes cada valor aparece
# Os resultados são armazenados nas variáveis "valores" e "contagem"

valor_mais_repetido = valores[np.argmax(contagem)]
#Usamos o "argmax" para encontrar a posição onde está a maior contagem
# Depois usamos essa posição para encontrar, dentro de "valores", o número que mais se repete

print("Array:", array)
print("Valor que mais se repete:", valor_mais_repetido)

#===============================================================================
#justificativa:

# Para descobrir qual número aparece mais vezes, utilizo o "unique()" junto
# com "return_counts=True", pois assim consigo obter os valores existentes
# e a quantidade de vezes que cada um aparece. Em seguida, uso "argmax()"
# para identificar qual dessas quantidades é a maior e localizar o valor
# correspondente no array.
```

## Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o angulo entre eles (produto escalar normalizado).

```python
import numpy as np

#criamos duas arrays com 20 números inteiros aleatórios entre 1 e 100

array1 = np.random.randint(1, 101, 20)
array2 = np.random.randint(1, 101, 20)

#para o tamanho, usamos umas funções para calcular o tamanho dos array's
#np.linalg.norm = calcula o tamanho de um vetor.
tamanho_array1 = np.linalg.norm(array1)
tamanho_array2 = np.linalg.norm(array2)

#aqui usamos np.dot para o calcular o produto interno entre os dois arrays
produto_escalar = np.dot(array1, array2)

#Arccos é uma função para o calcular o angulo entre os arrays e dividimos pelo o produto conforme a formula para encontrar o angulo entre os vetores
# Tecnicamente falando, o arccos calcula o cosseno de um valor
angulo_rad = np.arccos(produto_escalar / (tamanho_array1 * tamanho_array2))

#np.degrees é usado para o convereter o angulo dos arrays de radianos para graus.
# Tecnicamente falando, ele converte um angulo de radianos para graus

angulo_graus = np.degrees(angulo_rad)

print("Array 1:", array1)
print("Array 2:", array2)
print("Ângulo entre os arrays em radianos:", angulo_rad)
print("Ângulo entre os arrays em graus:", angulo_graus)

#===============================================================================
#justificativa:

# Para encontrar o ângulo entre os dois arrays, trato seus valores como vetores
# e utilizo a relação entre produto escalar e norma. Primeiro obtenho o tamanho
# de cada vetor e seu produto escalar para então aplicar a fórmula do cosseno.
# Depois, converto o resultado de radianos para graus para facilitar a
# interpretação do ângulo encontrado.
```

##  Calcule a integral da funcao f(x) = x² sin(x) no intervalo de 0 a π

```python
import numpy as np
from scipy.integrate import quad # Importamos o "quad" da biblioteca SciPy para realizar integrais numericamente

# Criamos a função f(x) = x² * sen(x)
def funcao(x):
    return x**2 * np.sin(x)

#QUAD é uma função do SciPy usada para calcular uma integral definida numericamente.
# Usamos o "quad" para calcular a integral da função no intervalo de 0 até pi
# O primeiro valor retornado é o resultado da integral e o segundo é uma estimativa do erro

integral, erro = quad(funcao, 0, np.pi)
#Como o quad retorna dois valores, usamos "integral, erro = quad(funcao, 0, np.pi)"
#A primeira variavel recebe o resultado integral e a segunda a estimativa de erro

print("Valor da integral:", integral)

#===============================================================================
#justificativa:

# Para calcular a integral definida da função, utilizo o "quad()" porque ele
# realiza a integração numericamente dentro do intervalo informado. Dessa
# forma, não preciso calcular a integral manualmente e ainda obtenho uma
# estimativa do erro junto com o resultado.
```

## Resolva a equacao diferencial dy/dx = x − y, com y(0) = 1, no intervalo de 0a 10.

# Use a funcao de otimizacao de SciPy para maximizar f(x) = −x 2+ 4x+ 1.


```python
from scipy.optimize import minimize_scalar # Importamos a função "minimize_scalar" do módulo "optimize" do SciPy para encontrar o menor valor de uma função

# Criamos a função f(x) = -x² + 4x + 1
def funcao(x):
    return -x**2 + 4*x + 1

# Como o "minimize_scalar" procura o menor valor, usamos o sinal negativo...
# para encontrar o maior valor da função
def funcao_negativa(x):
    return -funcao(x)

# Usamos o "minimize_scalar" para encontrar o valor de x que maximiza a função
resultado = minimize_scalar(funcao_negativa)

# Pegamos o valor de x encontrado pelo método de otimização
x_maximo = resultado.x

# Calculamos o valor máximo da função utilizando o x encontrado
valor_maximo = funcao(x_maximo)

print("Valor de x que maximiza a função:", x_maximo)

print("Valor máximo da função:", valor_maximo)

#===============================================================================
#justificativa:

# Para encontrar o maior valor da função, utilizo o "minimize_scalar()" sobre
# uma versão negativa da função. Isso acontece porque o método é desenvolvido
# para procurar valores mínimos. Ao inverter o sinal da função, o menor valor
# encontrado corresponde ao maior valor da função original. Depois, utilizo o
# valor de x encontrado para calcular o máximo na função original.
```

## Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).

```python
import numpy as np

from scipy.fft import fft, fftfreq # Importamos as funções "fft" e "fftfreq" do módulo "fft" do SciPy

# Criamos um intervalo de tempo entre 0 e 1 segundo com 1000 pontos
t = np.linspace(0, 1, 1000, endpoint=False)

# Criamos o sinal composto pelas funções sen(2πt) e cos(4πt)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

# Usamos o "fft" para calcular a Transformada Rápida de Fourier do sinal
transformada = fft(sinal)

# Calculamos as frequências correspondentes aos valores da transformada
frequencias = fftfreq(len(t), 1 / 1000)

# Calculamos a intensidade das frequências presentes no sinal
magnitude = np.abs(transformada)

print("Frequências:")

print(frequencias)

print("\n Magnitude da Transformada de Fourier:")

print(magnitude)

#===============================================================================
#justificativa:

# Para analisar quais frequências estão presentes no sinal, utilizo a
# Transformada Rápida de Fourier (FFT), pois ela permite representar o sinal
# no domínio das frequências. Utilizo o "fftfreq()" para associar cada valor
# da transformada à sua respectiva frequência e calculo a magnitude para
# identificar a intensidade de cada componente de frequência.
```

## Ajuste uma curva quadratica aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8)

```python
import numpy as np

from scipy.optimize import curve_fit # Importamos a função "curve_fit" do módulo "optimize" do SciPy para ajustar uma curva aos dados

# Criamos os valores de x dos pontos fornecidos
x = np.array([1, 2, 3, 4, 5])

# Criamos os valores de y dos pontos fornecidos
y = np.array([2, 4, 5, 7, 8])

# Criamos a função que representa uma curva quadrática
def funcao_quadratica(x, a, b, c):
    return a * x**2 + b * x + c

# Usamos o "curve_fit" para encontrar os valores de a, b e c que melhor ajustam a curva aos pontos
parametros, _ = curve_fit(funcao_quadratica, x, y)

# Armazenamos os valores encontrados para a, b e c
a, b, c = parametros

print("Valor de a:", a)

print("Valor de b:", b)

print("Valor de c:", c)

print("\nEquação da curva quadrática:")

print(f"y = {a:.2f}x² + {b:.2f}x + {c:.2f}")

#===============================================================================
#justificativa:

# Para encontrar uma curva que represente da melhor forma possível os pontos
# fornecidos, utilizo o "curve_fit()", pois ele ajusta os parâmetros de uma
# função aos dados informados. Escolho uma função quadrática porque o exercício
# pede esse tipo de curva, permitindo encontrar os valores de a, b e c que
# melhor se aproximam dos pontos de x e y.
```

## Filtre um DataFrame para exibir apenas os valores onde a coluna Preco e maior que 100.

## adicione uma coluna chamada Desconto ao DataFrame com 10% de desconto para todos os produtos

## Crie uma nova coluna chamada Total que seja o produto entre as colunas Preco e Quantidade

```python
import pandas as pd

dados = {
    "Produto": ["Arroz", "Feijão", "Macarrão", "Leite", "Café"],
    "Preco": [25.00, 8.50, 6.00, 5, 10],
    "Quantidade": [10, 15, 20, 12, 8]
}

dataframe = pd.DataFrame(dados)

# Criamos uma nova coluna chamada "Total" que é o produto entre as colunas "Preco" e "Quantidade"
dataframe["Total"] = dataframe["Preco"] * dataframe["Quantidade"]

print("DataFrame original:")

print(dataframe)

print("\nDataFrame com a nova coluna:")

print(dataframe)

#===============================================================================
#justificativa:

# Para calcular o valor total de cada produto, crio a coluna "Total" a partir
# da multiplicação entre "Preco" e "Quantidade". Utilizo essa operação
# diretamente entre as duas colunas porque o Pandas realiza o cálculo
# correspondente para cada linha automaticamente, evitando a necessidade de
# percorrer os produtos individualmente.
```

## Agrupe o DataFrame pela coluna Categoria e calcule o total de produtos por categoria

```python
import pandas as pd

# Criamos um DataFrame com produtos e suas respectivas categorias
dados = {
    "Produto": ["Arroz", "Feijão", "Macarrão", "Leite", "Café"],
    "Categoria": ["Alimentos", "Alimentos", "Alimentos", "Bebidas", "Bebidas"]
}

dataframe = pd.DataFrame(dados)

# Usamos o "groupby" para agrupar os produtos pela coluna "Categoria"
# Usamos o "count" para contar quantos produtos existem em cada categoria
produtos_por_categoria = dataframe.groupby("Categoria")["Produto"].count()

print("Total de produtos por categoria:")

print(produtos_por_categoria)

#===============================================================================
#justificativa:

# Para descobrir quantos produtos existem em cada categoria, utilizo o
# "groupby()", pois ele permite separar os produtos de acordo com a categoria
# à qual pertencem. Depois, utilizo o "count()" para contabilizar quantos
# produtos existem em cada grupo. Dessa forma, consigo obter uma contagem
# individual para cada categoria sem precisar realizar a contagem manualmente.
```

## Preencha os valores nulos de uma coluna num´erica com a mediana dessa coluna.

```python
import pandas as pd

dados = {
    "Coluna1": [1, 2, None, 4, 5],
    "Coluna2": [6, 7, 8, 9, 10]
}

dataframe = pd.DataFrame(dados)

# Calculamos a mediana da coluna "Coluna1"
mediana_coluna1 = dataframe["Coluna1"].median()

# Preenchemos os valores nulos da coluna "Coluna1"

dataframe["Coluna1"] = dataframe["Coluna1"].fillna(mediana_coluna1)

print("DataFrame original:")

print(dataframe)

print("\nDataFrame com os valores nulos preenchidos:")

print(dataframe)

#===============================================================================
#justificativa:

# Para tratar o valor ausente da coluna "Coluna1", utilizo a mediana como
# substituição, pois ela representa o valor central dos dados e é menos
# influenciada por valores muito altos ou muito baixos. Primeiro calculo a
# mediana da coluna e depois utilizo o "fillna()" para substituir o valor nulo,
# mantendo a quantidade de registros do DataFrame sem precisar remover a linha.
```

## Verifique quais colunas de um DataFrame contem valores duplicados

```python
import pandas as pd

dados = {
    "Coluna1": [1, 2, 2, 4, 5],
    "Coluna2": [6, 7, 7, 9, 10],
    "Coluna3": ["A", "B", "D", "F", "E"]
}

dataframe = pd.DataFrame(dados)

# Verificamos quais colunas possuem valores duplicados
duplicados = dataframe.apply(lambda coluna: coluna.duplicated().any())

print("Colunas com valores duplicados:")

print(duplicados)

#===============================================================================
#justificativa:

# Para identificar quais colunas possuem valores repetidos, utilizo o
# "duplicated()", pois ele permite verificar a ocorrência de duplicatas em
# cada coluna. Utilizo o "apply()" para aplicar essa verificação
# individualmente em todas as colunas e o "any()" para indicar se existe pelo
# menos um valor duplicado em cada uma delas.
```

## Crie uma tabela dinˆamica a partir do DataFrame, calculando a soma de Quantidade por Categoria

```python
import pandas as pd

dados = {
    "Produto": ["Arroz", "Feijão", "Macarrão", "Leite", "Café"],
    "Categoria": ["Alimentos", "Alimentos", "Alimentos", "Bebidas", "Bebidas"],
    "Quantidade": [10, 15,20,12,8]
}

dataframe = pd.DataFrame(dados)

# Usamos o "pivot_table" para criar uma tabela dinâmica
tabela_dinamica = pd.pivot_table(dataframe, values="Quantidade", index="Categoria", aggfunc="sum")

print("DataFrame original:")

print(dataframe)

print("\nTabela dinâmica:")

print(tabela_dinamica)

#===============================================================================
#justificativa:

# Para resumir a quantidade de produtos de cada categoria, utilizo o
# "pivot_table()", pois ele permite organizar e agrupar os dados em uma tabela
# dinâmica. Defino "Categoria" como índice para separar os produtos por grupo,
# "Quantidade" como o valor que será analisado e utilizo o "sum" para obter a
# soma das quantidades dentro de cada categoria.
```

## Mescle dois DataFrames baseados em uma coluna comum chamada ID utilizando um join.

```python
import pandas as pd

dados1 = {
    "ID": [1, 2, 3, 4],
    "Produto": ["Arroz", "Feijão", "Macarrão", "Leite"]
}

dataframe1 = pd.DataFrame(dados1)

dados2 = {
    "ID": [1, 2, 3, 4],
    "Preco": [25.00, 8.50, 6.00, 5.00]
}

dataframe2 = pd.DataFrame(dados2)

# Transformamos a coluna "ID" em índice nos dois DataFrames
dataframe1 = dataframe1.set_index("ID")
dataframe2 = dataframe2.set_index("ID")

# Usamos o "join" para juntar os dois DataFrames através do índice "ID"
dataframe_mesclado = dataframe1.join(dataframe2)

print("DataFrame mesclado:")

print(dataframe_mesclado)

#===============================================================================
#justificativa:

# Para combinar as informações dos dois DataFrames, utilizo o "join()", pois
# ele permite unir tabelas que possuem um índice em comum. Transformo a coluna
# "ID" em índice nos dois DataFrames para que o "join()" possa relacionar cada
# produto ao seu respectivo preço de acordo com o mesmo identificador.
```

## Crie um DataFrame com dados faltantes e use a interpolacao linear para preenche-los.

```python
import pandas as pd

dados = {
    "Valor": [10, None, 30, None, 50]
}

dataframe = pd.DataFrame(dados)

# Usamos o "interpolate" para preencher os valores faltantes utilizando interpolação linear
dataframe["Valor"] = dataframe["Valor"].interpolate()

print("DataFrame original:")

print(dataframe)

print("\nDataFrame com os valores faltantes preenchidos:")

print(dataframe)

#===============================================================================
#justificativa:

# Para preencher os valores ausentes, utilizo o "interpolate()", pois ele
# permite estimar os valores faltantes com base nos valores que estão ao redor.
# Utilizo a interpolação linear porque os dados possuem uma sequência numérica,
# permitindo calcular um valor intermediário entre os registros conhecidos sem
# precisar remover as linhas que possuem valores ausentes.
```

## Adicione uma coluna ao DataFrame que calcule o imposto (5%) sobre o valor da coluna Total.

```python
import pandas as pd

dados = {
    "Produto": ["Arroz", "Feijão", "Macarrão", "Leite", "Café"],
    "Preco": [25.00, 8.50, 6.00, 5.00, 10.00],
    "Quantidade": [10, 15, 20, 12, 8]
}

dataframe = pd.DataFrame(dados)

# Criamos uma nova coluna chamada "Total" que é o produto entre as colunas "Preco" e "Quantidade"
dataframe["Total"] = dataframe["Preco"] * dataframe["Quantidade"]

# Criamos uma nova coluna chamada "Imposto" que é o imposto de 5% sobre a coluna "Total"
dataframe["Imposto"] = dataframe["Total"] * 0.05

print("DataFrame original:")

print(dataframe)

print("\nDataFrame com a nova coluna:")

print(dataframe)

#===============================================================================
#justificativa:

# Para calcular o imposto de cada produto, utilizo primeiro a coluna "Total",
# pois o imposto de 5% deve ser aplicado sobre o valor total da compra e não
# somente sobre o preço de uma unidade. Multiplico o "Total" por 0.05, pois
# esse valor representa 5%, e armazeno o resultado em uma nova coluna chamada
# "Imposto", permitindo manter os valores originais e o imposto separados.
```

## Plote dois gr´aficos de linha em subplots diferentes: y1 = e x e y2 = log(x).

```python
import numpy as np

import matplotlib.pyplot as plt #Usamos o matplotlib e batizamos o pyplot para "plt" para facilitar o uso de suas funções

#Criamos valores de x começando em 0.1 até 5
#Começamos em 0.1 porque a função log(x) não aceita valores iguais ou menores que 0

x = np.linspace(0.1, 5, 100)

#Calculamos os valores de y1 utilizando a função exponencial e^x
#Usamos o "np.exp" para calcular a exponencial

y1 = np.exp(x)

#Calculamos os valores de y2 utilizando a função log(x)
#Usamos o "np.log" para calcular o logaritmo natural

y2 = np.log(x)

#Usamos o "subplots(1, 2)" para criar 1 linha com 2 gráficos diferentes
#O "fig" representa a figura completa e "ax" representa os dois espaços onde os gráficos serão desenhados

fig, ax = plt.subplots(1, 2)

#Usamos o primeiro espaço "ax[0]" para criar o gráfico da função y1 = e^x

ax[0].plot(x, y1)

#Adicionamos um título ao primeiro gráfico

ax[0].set_title("y1 = e^x")

#Usamos o segundo espaço "ax[1]" para criar o gráfico da função y2 = log(x)

ax[1].plot(x, y2)

#Adicionamos um título ao segundo gráfico

ax[1].set_title("y2 = log(x)")

#Ajustamos automaticamente o espaço entre os dois gráficos para evitar que os elementos fiquem sobrepostos

plt.tight_layout()

#Por fim, usamos o "show" para exibir os dois gráficos

plt.show()

#===============================================================================
#justificativa:

# Para comparar visualmente as funções exponencial e logarítmica, utilizo
# "subplots(1, 2)", pois ele permite apresentar os dois gráficos dentro da
# mesma figura, facilitando a comparação entre eles. Também utilizo o mesmo
# conjunto de valores de x para que as duas funções sejam analisadas no mesmo
# intervalo. O intervalo começa em 0.1 porque o logaritmo natural só pode ser
# calculado para valores maiores que zero.
```

## Plote um grafico de linha com intervalos de confianca (sombramento) para a funcao y = sin(x).

```python
import numpy as np

import matplotlib.pyplot as plt

#Criamos 100 valores de x distribuídos entre 0 e 10

x = np.linspace(0, 10, 100)

#Calculamos os valores de y utilizando a função seno

y = np.sin(x)

#Criamos o limite inferior do intervalo de confiança
#Subtraímos 0.2 dos valores de y para criar uma margem abaixo da função

y_inferior = y - 0.2

#Criamos o limite superior do intervalo de confiança
#Adicionamos 0.2 aos valores de y para criar uma margem acima da função

y_superior = y + 0.2

#Usamos o "plot" para criar o gráfico de linha da função seno

plt.plot(x, y, label="sin(x)")

#Usamos o "fill_between" para preencher o espaço entre o limite inferior e o limite superior
#Esse preenchimento cria visualmente o intervalo de confiança ao redor da linha

plt.fill_between(x, y_inferior, y_superior, alpha=0.2, label="Intervalo de confiança")


plt.title("Função seno com intervalo de confiança")


plt.xlabel("x")


plt.ylabel("y")


plt.legend()

#Por fim, usamos o "show" para exibir o gráfico

plt.show()

#===============================================================================
#justificativa:

# Para representar a relação entre os valores de x e y, utilizo o "scatter()",
# pois esse tipo de gráfico permite visualizar cada par de valores como um
# ponto separado. Também utilizo o parâmetro "s" para variar o tamanho dos
# pontos de acordo com os valores de y, permitindo representar uma informação
# adicional no próprio gráfico e tornando as diferenças entre os valores mais
# perceptíveis.
```

## Crie um grafico de barras empilhadas para tres categorias com tres grupos de valores.

```python
import matplotlib.pyplot as plt

categorias = ["Categoria A", "Categoria B", "Categoria C"]

grupo1 = [10, 20, 15]
grupo2 = [15, 10, 20]
grupo3 = [5, 15, 10]

#Usamos o "bar" para criar as primeiras barras do gráfico

plt.bar(categorias, grupo1, label="Grupo 1")

#-------------------------------------------------------------------------------
plt.bar(categorias, grupo2, bottom=grupo1, label="Grupo 2")
#Usamos novamente o "bar" para criar o segundo grupo
#O parâmetro "bottom" indica que o segundo grupo deve começar acima dos valores do primeiro grupo
#Assim, os valores dos dois grupos ficam empilhados
#-------------------------------------------------------------------------------

#Criamos o terceiro grupo utilizando novamente o "bottom"
#Aqui usamos a soma de "grupo1" e "grupo2" para que o terceiro grupo comece acima dos dois anteriores

plt.bar(categorias, grupo3,
        bottom=[grupo1[i] + grupo2[i] for i in range(len(categorias))],
        label="Grupo 3")


plt.title("Gráfico de barras empilhadas")


plt.xlabel("Categorias")


plt.ylabel("Valores")


plt.legend()


plt.show()

#===============================================================================
#justificativa:

# Para comparar a composição dos três grupos dentro de cada categoria, utilizo
# um gráfico de barras empilhadas. O parâmetro "bottom" permite posicionar cada
# grupo acima do anterior, fazendo com que os valores sejam apresentados de
# forma acumulada. Dessa maneira, consigo visualizar tanto a contribuição de
# cada grupo quanto o total representado em cada categoria. Utilizo o
# "legend()" para identificar qual parte das barras pertence a cada grupo.
```

## Gere um grafico de dispersao 3D para 100 pontos aleatorios utilizando Matplotlib.

```python
import numpy as np

import matplotlib.pyplot as plt

#Uso o "random.rand" para gerar 100 valores aleatórios para cada eixo
#Os valores gerados ficam entre 0 e 1
x = np.random.rand(100)
y = np.random.rand(100)
z = np.random.rand(100)

#Criamos uma figura para receber o nosso gráfico
fig = plt.figure()

#Uso o "add_subplot" para adicionar um espaço para o gráfico
#O "111" representa uma figura com 1 linha, 1 coluna e o primeiro gráfico
#O parâmetro "projection='3d'" transforma o espaço em um gráfico tridimensional

ax = fig.add_subplot(111, projection="3d")

#Uso o "scatter" para criar o gráfico de dispersão 3D
#Os valores de x, y e z representam as posições dos pontos nos três eixos

ax.scatter(x, y, z)


ax.set_title("Gráfico de dispersão 3D")


ax.set_xlabel("X")


ax.set_ylabel("Y")


ax.set_zlabel("Z")

plt.show()

#===============================================================================
#justificativa:

# Para representar pontos considerando três dimensões, utilizo um gráfico de
# dispersão 3D, pois ele permite visualizar simultaneamente as relações entre
# os eixos X, Y e Z. Utilizo o "add_subplot()" com "projection='3d'" para criar
# o espaço tridimensional e o "scatter()" para posicionar cada ponto de acordo
# com seus respectivos valores nos três eixos.
```
