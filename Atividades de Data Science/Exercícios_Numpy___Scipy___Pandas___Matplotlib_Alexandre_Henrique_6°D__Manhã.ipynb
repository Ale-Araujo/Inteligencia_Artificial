## Crie dois arrays de numeros inteiros aleatorios de tamanho 20 e calcule o angulo entre eles (produto escalar normalizado).

```python
import numpy as np

#criamos duas arrays com 20 números inteiros aleatórios entre 1 e 100

array1 = np.random.randint(1, 101, 20)
array2 = np.random.randint(1, 101, 20)

#para o tamanho, usamos umas funções para calcular o tamanho dos array's
#np.linalg.norm = calcula o tamanho de um vetor.
tamanho_array1 = np.linalg.norm(array1)
tamanho_array2 = np.linalg.norm(array2)

#aqui usamos np.dot para o calcular o produto interno entre os dois arrays
produto_escalar = np.dot(array1, array2)

#Arccos é uma função para o calcular o angulo entre os arrays e dividimos pelo o produto conforme a formula para encontrar o angulo entre os vetores
# Tecnicamente falando, o arccos calcula o cosseno de um valor
angulo_rad = np.arccos(produto_escalar / (tamanho_array1 * tamanho_array2))

#np.degrees é usado para o convereter o angulo dos arrays de radianos para graus.
# Tecnicamente falando, ele converte um angulo de radianos para graus

angulo_graus = np.degrees(angulo_rad)

print("Array 1:", array1)
print("Array 2:", array2)
print("Ângulo entre os arrays em radianos:", angulo_rad)
print("Ângulo entre os arrays em graus:", angulo_graus)

```

##  Calcule a integral da funcao f(x) = x² sin(x) no intervalo de 0 a π

```python
import numpy as np
from scipy.integrate import quad # Importamos o "quad" da biblioteca SciPy para realizar integrais numericamente

# Criamos a função f(x) = x² * sen(x)
def funcao(x):
    return x**2 * np.sin(x)

#QUAD é uma função do SciPy usada para calcular uma integral definida numericamente.
# Usamos o "quad" para calcular a integral da função no intervalo de 0 até pi
# O primeiro valor retornado é o resultado da integral e o segundo é uma estimativa do erro

integral, erro = quad(funcao, 0, np.pi)
#Como o quad retorna dois valores, usamos "integral, erro = quad(funcao, 0, np.pi)"
#A primeira variavel recebe o resultado integral e a segunda a estimativa de erro

print("Valor da integral:", integral)
```

## Resolva a equacao diferencial dy/dx = x − y, com y(0) = 1, no intervalo de 0a 10.

```python
import numpy as np # Usamos o numpy e o batizamos de "np" para facilitar o uso de suas funções

from scipy.integrate import solve_ivp # Importamos a função "solve_ivp" do módulo "integrate" do SciPy para resolver equações diferenciais

# Criamos a função que representa a equação diferencial dy/dx = x - y
def equacao(x, y):
    return x - y

# Definimos o intervalo de x entre 0 e 10
intervalo = (0, 10)

# Definimos a condição inicial y(0) = 1
condicao_inicial = [1]

# Usamos o "solve_ivp" para resolver numericamente a equação diferencial
solucao = solve_ivp(equacao, intervalo, condicao_inicial)

print("Valores de x:")

print(solucao.t)

# O ".t" apresenta os valores de x utilizados durante a resolução da equação

print("\nValores de y:")

print(solucao.y[0])

# O ".y" apresenta os valores de y encontrados para cada valor de x
# O "[0]" seleciona os valores da primeira variável da solução
```

## Encontre as raızes da funcao f(x) = x 4 − 3x³ + 2 utilizando SciPy

```python
from scipy.optimize import fsolve # Importamos a função "fsolve" do módulo "optimize" do SciPy para encontrar raízes de funções

# Criamos a função f(x) = x⁴ - 3x³ + 2
def funcao(x):
    return x**4 - 3*x**3 + 2

# Usamos o "fsolve" para encontrar as raízes da função


raiz1 = fsolve(funcao, -1)

raiz2 = fsolve(funcao, 1)

raiz3 = fsolve(funcao, 3)

# Os valores "-1", "1" e "3" são valores iniciais utilizados para encontrar as raízes

print("Raízes da função:")

print(raiz1)

print(raiz2)

print(raiz3)
```

# Use a funcao de otimizacao de SciPy para maximizar f(x) = −x 2+ 4x+ 1.


```python
from scipy.optimize import minimize_scalar # Importamos a função "minimize_scalar" do módulo "optimize" do SciPy para encontrar o menor valor de uma função

# Criamos a função f(x) = -x² + 4x + 1
def funcao(x):
    return -x**2 + 4*x + 1

# Como o "minimize_scalar" procura o menor valor, usamos o sinal negativo...
# para encontrar o maior valor da função
def funcao_negativa(x):
    return -funcao(x)

# Usamos o "minimize_scalar" para encontrar o valor de x que maximiza a função
resultado = minimize_scalar(funcao_negativa)

# Pegamos o valor de x encontrado pelo método de otimização
x_maximo = resultado.x

# Calculamos o valor máximo da função utilizando o x encontrado
valor_maximo = funcao(x_maximo)

print("Valor de x que maximiza a função:", x_maximo)

print("Valor máximo da função:", valor_maximo)
```

## Calcule a transformada de Fourier de um sinal composto de sin(2πt) e cos(4πt).

```python
import numpy as np

from scipy.fft import fft, fftfreq # Importamos as funções "fft" e "fftfreq" do módulo "fft" do SciPy

# Criamos um intervalo de tempo entre 0 e 1 segundo com 1000 pontos
t = np.linspace(0, 1, 1000, endpoint=False)

# Criamos o sinal composto pelas funções sen(2πt) e cos(4πt)
sinal = np.sin(2 * np.pi * t) + np.cos(4 * np.pi * t)

# Usamos o "fft" para calcular a Transformada Rápida de Fourier do sinal
transformada = fft(sinal)

# Calculamos as frequências correspondentes aos valores da transformada
frequencias = fftfreq(len(t), 1 / 1000)

# Calculamos a intensidade das frequências presentes no sinal
magnitude = np.abs(transformada)

print("Frequências:")

print(frequencias)

print("\n Magnitude da Transformada de Fourier:")

print(magnitude)
```

## Use a funcao de interpolacao cubica de SciPy para interpolar os dados (0, 0),(1, 2),(2, 3),(3, 5),(4, 4).

## Ajuste uma curva quadratica aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8)

```python
import numpy as np

from scipy.optimize import curve_fit # Importamos a função "curve_fit" do módulo "optimize" do SciPy para ajustar uma curva aos dados

# Criamos os valores de x dos pontos fornecidos
x = np.array([1, 2, 3, 4, 5])

# Criamos os valores de y dos pontos fornecidos
y = np.array([2, 4, 5, 7, 8])

# Criamos a função que representa uma curva quadrática
def funcao_quadratica(x, a, b, c):
    return a * x**2 + b * x + c

# Usamos o "curve_fit" para encontrar os valores de a, b e c que melhor ajustam a curva aos pontos
parametros, _ = curve_fit(funcao_quadratica, x, y)

# Armazenamos os valores encontrados para a, b e c
a, b, c = parametros

print("Valor de a:", a)

print("Valor de b:", b)

print("Valor de c:", c)

print("\nEquação da curva quadrática:")

print(f"y = {a:.2f}x² + {b:.2f}x + {c:.2f}")
```

## Utilize a funcao de otimizacao multivariada para minimizar f(x, y)= (x−2)2 + (y − 3)2.


```python
from scipy.optimize import minimize # Importamos a função "minimize" do módulo "optimize" do SciPy para minimizar funções

# Criamos a função f(x, y) = (x - 2)² + (y - 3)²
def funcao(variaveis):
    x, y = variaveis
    return (x - 2)**2 + (y - 3)**2

# Definimos os valores iniciais de x e y
valor_inicial = [0, 0]

# Usamos o "minimize" para encontrar os valores de x e y que minimizam a função
resultado = minimize(funcao, valor_inicial)

print("Valor de x:", resultado.x[0])

print("Valor de y:", resultado.x[1])

print("\nValor mínimo da função:")

print(resultado.fun)
```

## Calcule a transformada de Fourier inversa de um sinal complexo gerado aleatoriamente

```python
import numpy as np # Usamos o numpy e o batizamos de "np" para facilitar o uso de suas funções

from scipy.fft import ifft # Importamos a função "ifft" do módulo "fft" do SciPy para calcular a Transformada de Fourier inversa

# Criamos um sinal complexo com números aleatórios
sinal = np.random.rand(10) + 1j * np.random.rand(10)

# Usamos o "ifft" para calcular a Transformada de Fourier inversa do sinal
sinal_inverso = ifft(sinal)

print("Sinal complexo:")

print(sinal)

print("\nTransformada de Fourier inversa:")

print(sinal_inverso)
```

## Resolva a equacao diferencial de segunda ordem d2y dx2 +y = 0, com condicoes iniciais y(0) = 1 e y′(0) = 0.

```python
import numpy as np # Usamos o numpy e o batizamos de "np" para facilitar o uso de suas funções

from scipy.integrate import solve_ivp # Importamos a função "solve_ivp" do módulo "integrate" do SciPy para resolver equações diferenciais

# Criamos a função que representa a equação diferencial y'' + y = 0
def equacao(x, y):
    return [y[1], -y[0]]

# Definimos o intervalo de x entre 0 e 10
intervalo = (0, 10)

# Definimos as condições iniciais y(0) = 1 e y'(0) = 0
condicao_inicial = [1, 0]

# Usamos o "solve_ivp" para resolver numericamente a equação diferencial
solucao = solve_ivp(equacao, intervalo, condicao_inicial)

print("Valores de x:")

print(solucao.t)

print("\nValores de y:")

print(solucao.y[0])
```

## Interpole os valores de uma funcao f(x) = x2 nos pontos [1, 2, 3, 4, 5] utilizando spline cubica.

```python
import numpy as np # Usamos o numpy e o batizamos de "np" para facilitar o uso de suas funções

from scipy.interpolate import CubicSpline # Importamos a função "CubicSpline" do módulo "interpolate" do SciPy para realizar interpolação por spline cúbica

# Criamos os pontos de x fornecidos
x = np.array([1, 2, 3, 4, 5])

# Calculamos os valores de y utilizando a função f(x) = x²
y = x**2

# Usamos o "CubicSpline" para criar uma função de interpolação por spline cúbica
spline = CubicSpline(x, y)

# Criamos novos valores de x entre 1 e 5 para realizar a interpolação
novos_x = np.linspace(1, 5, 20)

# Calculamos os valores interpolados de y para os novos valores de x
novos_y = spline(novos_x)

print("Valores originais de x:")

print(x)

print("\nValores originais de y:")

print(y)

print("\nValores interpolados:")

print(novos_y)
```

## Realize a transformada de Fourier de um sinal discreto composto de senos e cossenos.

```python
import numpy as np # Usamos o numpy e o batizamos de "np" para facilitar o uso de suas funções

from scipy.fft import fft # Importamos a função "fft" do módulo "fft" do SciPy para calcular a Transformada de Fourier

# Criamos 100 pontos discretos
n = np.arange(100)

# Criamos um sinal composto por seno e cosseno
sinal = np.sin(2 * np.pi * 5 * n / 100) + np.cos(2 * np.pi * 10 * n / 100)

# Usamos o "fft" para calcular a Transformada de Fourier do sinal
transformada = fft(sinal)

# Calculamos a intensidade das frequências presentes no sinal
magnitude = np.abs(transformada)

print("Sinal discreto:")

print(sinal)

print("\nTransformada de Fourier:")

print(transformada)

print("\nMagnitude da Transformada:")

print(magnitude)
```

## Resolva um sistema de equacoes lineares com tres variaveis utilizando SciPy.


```python
import numpy as np # Usamos o numpy e o batizamos de "np" para facilitar o uso de suas funções

from scipy.linalg import solve # Importamos a função "solve" do módulo "linalg" do SciPy para resolver sistemas de equações lineares

# Criamos a matriz com os coeficientes das três equações
matriz = np.array([[1, 1, 1],
                   [2, 1, -1],
                   [1, 2, 1]])

# Criamos o array com os resultados das três equações
resultado = np.array([6, 3, 8])

# Usamos o "solve" para encontrar os valores de x, y e z
solucao = solve(matriz, resultado)

print("Valor de x:", solucao[0])

print("Valor de y:", solucao[1])

print("Valor de z:", solucao[2])
```
